# 1(b) Smallest even degree $m$ with discretization error $\le 10^{-6}$ for $P_m(1)$

The Taylor polynomial of $\cos(x)$ about $x=0$ truncated at degree $m=2n$ is

$$P_{2n}(x)=\sum_{k=0}^{n}\frac{(-1)^k}{(2k)!}x^{2k}.$$

The **discretization (truncation) error** at $x=1$ is $E_m=\left|\cos(1)-P_m(1)\right|$.
We add one term at a time until $E_m\le 10^{-6}$.

```python
from math import factorial

COS1 = 0.5403023058681397  # exact value of cos(1) given in the problem
TOL = 1e-6


def P(m, x):
    """Taylor polynomial of cos(x) about 0, truncated at even degree m."""
    return sum((-1) ** k * x ** (2 * k) / factorial(2 * k) for k in range(m // 2 + 1))


def smallest_even_degree(x=1.0, tol=TOL, exact=COS1, max_m=200):
    """Smallest even m with |exact - P_m(x)| <= tol. Returns (m, P_m(x), error, table)."""
    table = []
    partial = 0.0
    for k in range(max_m // 2 + 1):
        m = 2 * k
        partial += (-1) ** k * x ** (2 * k) / factorial(2 * k)  # add the degree-m term
        error = abs(exact - partial)
        table.append((m, partial, error))
        if error <= tol:
            return m, partial, error, table
    raise RuntimeError(f"tolerance {tol} not met for m <= {max_m}")


m, value, error, table = smallest_even_degree()

print(f"{'m':>3} {'P_m(1)':>20} {'|cos(1) - P_m(1)|':>22}")
print('-' * 47)
for mm, pv, er in table:
    flag = '  <-- first m with error <= 1e-6' if er <= TOL else ''
    print(f"{mm:>3} {pv:>20.16f} {er:>22.6e}{flag}")

print()
print(f"Smallest even degree m = {m}")
print(f"P_{m}(1)               = {value:.16f}")
print(f"cos(1)                = {COS1:.16f}")
print(f"Discretization error   = {error:.6e}  (<= 1e-6)")
```

```python
# Sanity check: the previous even degree must NOT meet the tolerance.
prev_err = abs(COS1 - P(m - 2, 1.0))
print(f"error at m-2 = {m - 2}: {prev_err:.6e}  ->  {'meets' if prev_err <= TOL else 'exceeds'} 1e-6")
assert prev_err > TOL and abs(COS1 - P(m, 1.0)) <= TOL
print(f"Confirmed: m = {m} is the smallest even degree.")
```

**Answer:** $m = 8$, i.e. $P_8(1)=1-\tfrac12+\tfrac1{24}-\tfrac1{720}+\tfrac1{40320}\approx 0.5403025793650793$, with discretization error $\approx 2.735\times10^{-7}\le 10^{-6}$. The previous even degree $m=6$ gives $\approx 2.453\times10^{-5}$, which is too large.
